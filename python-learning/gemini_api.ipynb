```python
import requests

class GoogleGeminiLLM:
    def __init__(self, api_key, model="gemini-1.5-flash", temperature=0.7):
        self.api_key = api_key
        self.model = model
        self.temperature = temperature
        self.endpoint = f"https://generativelanguage.googleapis.com/v1beta2/models/{self.model}:generateMessage"

    def generate(self, prompt):
        headers = {"Content-Type": "application/json"}
        params = {"key": self.api_key}
        data = {
            "messages": [
                {"author": "user", "content": prompt}
            ],
            "temperature": self.temperature,
        }
        response = requests.post(self.endpoint, headers=headers, params=params, json=data)

        if response.status_code == 200:
            return response.json()["candidates"][0]["content"]
        else:
            raise Exception(f"Error: {response.status_code}, {response.text}")

# Configure the Google Gemini LLM
api_key = "AIzaSyCdn9Ye7uUmTjKJGwP172oGCdVgxFAGF4A"  # Replace with your actual API key
llm = GoogleGeminiLLM(api_key=api_key)

# Define a question
question = "What is LangChain?"
prompt = f"You are a helpful assistant. Answer the following question:\n\n{question}"

# Get the response
try:
    response = llm.generate(prompt)
    print("Answer:", response)
except Exception as e:
    print("Error:", e)
```

```python
from langchain.llms import GoogleGeminiFlash
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

# Configure the Gemini Flash model
llm = GoogleGeminiFlash(
    api_key="AIzaSyCdn9Ye7uUmTjKJGwP172oGCdVgxFAGF4A",
    model="gemini-flash",
    temperature=0.7  # Adjust for creativity
)

# Create a prompt template
prompt_template = PromptTemplate(
    input_variables=["question"],
    template="You are a helpful assistant. Answer the following question:\n\n{question}"
)

# Create the LLM chain
chain = LLMChain(llm=llm, prompt=prompt_template)

# Run the chain with a sample question
question = "What is LangChain?"
response = chain.run({"question": question})

print("Answer:", response)
```
